### 로우(행) 선택

````python
# 인텍스 이름을 기준
df.loc[0]
````

````python
# 위치번호 기준. 인텍스를 다른 값으로 바꿨을때 사용하는 방법
df.iloc[0]
````

````python
# 시리즈의 인덱스로 각 열의 값을 확인
df.loc[0]['major']
````

````python
# 슬라이싱으로 여러행을 선택.
# 파이썬 슬라이싱 규칙에서 벗어남. 뒤에 수가 n-1이 아님!!!
df.loc[0:2]
````

````python
#파이썬 배열 인덱스와 동일
df.iloc[0:2]
````

##### 행 선택, 인덱스행(loc)과 위치값행(iloc)차이

````python
df2 = pd.DataFrame(raw_data,
                    index=['5001','5002','5003','5004','5009','5010'])
df2
````

- loc는 부여받느 인덱스로 행선택, 슬라이싱
- iloc는 원래부터의 위치값(파이썬 배열의 인덱스)와 동일

````python
df.loc[0, 'name']
````

````python
# 인덱스를 따로 부여하지 않은 df에서 행과 열 선택
df.loc[0:2,['name','age']]
````

````python
# 인덱스를 따로 지정해준 df에서 행과 열 선택
df3 = df2.loc['5001':'5003',['name','age']]
df3 # 새 dataframe 생성
````

#### 조건 조회
- SQL WHERE 절과 동일

````python

df['score'] >= 90
````

````python
df[df['score'] >= 90]
````

- NumPy와 동일한 조건연산
- 파이썬 : and, or, not 사용
- NumPy, Pandas `:&,` `|,` `~` 사용
- 연산자 우선순위로 각 조건은 받느시 ()를 사용할 것

````python
# 점수가 90점 초과이고 나이가 21살 이상인 학생 데이터 참/거짓 여부
(df['score'] > 90) & (df['age'] >= 21)
````

````python
df[(df['score'] > 90) & (df['age'] >= 21)]
````

````python
# 점수가 90 초과 이거나 나이가 21살 이상이거나 둘 중에 하나로 참인 데이터 조회
df[(df['score'] > 90) | (df['age'] >= 21)]
````

#### 컬럼 추가

````python
df['score_bonus'] = df['score_bonus'] -2
df
````

### 실습

-상품데이터 활용

````python
data = {
    'product': ['키보드','마우스', '모니터', 'usb', '노트북'],
    'category': ['주변긱', '주변기기', '디스플레이', '저장장치', '컴퓨터'],
    'price': [30000, 15000, 25000, 12000, 120000],
    'stock': [10, 30, 5, 50, 3]
}

df_products = pd.DataFrame(data)
````

1. 전체 데이터 출력
2. product. price 컬럼 출력
3. 가격이 50000원 이상인 상품만 출력
4. 재고가 10개 이상인 상품만 출력
5. total_value 컬럼 추가. 계산식은 price * stock
6. 전체 재고의 합계 출력
7. 카테고리가 주변기기인 상품만 출력

````python
# 1번 
df_products
````

````python
# 2번
df_products[['product', 'price']]
````

````python
# 3번
df_products['price'] >= 50000
````

````python
df_products[df_products['price'] >= 50000]
````

````python
# 4번
df_products[df_products['stock'] >= 10]
````

````python
#5번 토탈벨류를 만들어라
df_products['total_value'] = df_products['price'] * df_products['stock']

df_products
````

````python
#6번 전체재고가치 합계출력
df_products['total_value'].sum()
````

````python
# 7번
df_products[df_products['category'] == '주변기기']
````
