#### 파이썬으로 배열작업

````python
prices = [10000, 15000, 12000, 30000, 25000]  # 5개 가격이 든 배열

discount_prices = []

for price in prices: 
    discount_prices.append(price * 0.9) # 10% 할인

discount_prices # 노트북 코드쉐에서 하나만 출력하려면 print() 생략가능
````

- 결측치를 정리방법 : 결측치를 제거하거나(건수가 줄어듬), 나머지 평균값을 임의로 기입하거나, 기본통계에서 도출할수 있는 다른 값들로 채우기
 

#### 이상치 핸들링
- 일반적인 범위를 벗어나는 비정상 데이터
- 예로 부산 1월 일별 평균기온에 -90도. 습도 99%ㅠ



````python
temperatures = np.array([22.5, 23.0, np.nan, 25.1, 999.0, 21.8, -99.0])
temperatures
````

````python
valid_temps = temperatures[
    (~np.isnan(temperatures) #nan은 무조건제거
      & (temperatures >= 15.0) #최저기온이 15 아래로 내려갈 수 없는처럼 제거
      &(temperatures <= 50) #최고기온이 50 이상 오를 수 없으므로 제거
]

valid_temps
````

- 전처리(preprocessing) - 분석 이전에 이상치, 결측치를 없애는 작업

#### 매출 데이터 전처리
- 결측값 개수 출력
- 결측값 제외
- 0원 10만원 내의 정상 매출만 남기고
- 정상 매출의 평균, 최대값, 최소값 출력
- 정상 매출에서 평균이상인 값만 출력

````python
raw_sales = [12000, 15000, None, 30000, -5000, 25000, 999999, 18000,None, 22000]


raw_sales
````

````python
# 파이썬 리스트에서 None대신 np.nan으로 결측치를 입력하면 NumPy 패키지 연산이 가능
raw_sales = [12000, 15000, np.nan, 30000, -5000, 25000, 999999, 18000, np.nan, 22000]


raw_sales
````

````python
# NumPy 배영로 변경
np_raw_sales = np.array(raw_sales)

np_raw_sales
````

````python
# 결측값 확인
np.isnan(np_raw_sales)
````

````python
# 결측값 확인
np.isnan(np_raw_sales).sum()
````

````python
np_raw_sales[
    (~np.isnan(np_raw_sales)) &
    (np_raw_sales >= 0) &
    (np_raw_sales <= 100000)
]

````

````python
valid_sales =np_raw_sales[
    (~np.isnan(np_raw_sales)) &
    (np_raw_sales >= 0) &
    (np_raw_sales <= 100000)
]

valid_sales
````

````python
# 평균 매출액 20333.3원
mean_sales = valid_sales.mean()
mean_sales
````

````python
valid_sales.min()
````

````python
valid_sales.max()
````

````python
# 평균 매출액보다 높은 매출 찾기
valid_sales[valid_sales > mean_sales]
````
