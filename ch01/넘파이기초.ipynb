### Python 리스트(배열) NumPy 배열 비교

- 파이썬에 기존 리스트(배열) 자료구조 존재, 속도적면에서 NumPy가 우세

#### 파이썬으로 배열작업

````python
prices = [10000, 15000, 12000, 30000, 25000]  # 5개 가격이 든 배열

discount_prices = []

for price in prices: 
    discount_prices.append(price * 0.9) # 10% 할인

discount_prices # 노트북 코드쉐에서 하나만 출력하려면 print() 생략가능
````

#### 넘파이로 배열작업

````python
np_prices = np.array(prices) #넘파이 배열로 변경
np_discount_prices = np_prices * 0.9

np_discount_prices
````

#### NumPy 데이터타딥
- int8 : -128 ~ 127 (합 256)
- int16 : -32768 ~ 32767 (총 65536)
- int32 : 32비트 정수
- int64 : 64비트 정수
- uint8 : 8 ~ 255  음수 없는 정수
- nint16 : 0 ~ 65535
- float16 : 실수
- float32 : 32비트 실수
- float64 : 64비트 실수(Numpy 기본 실수)
- complex64 : 복소수
- complex128 : 128비트 복소수
- bool : True/False
- str_ : 문자열
- object_ : 객체

#### 조건 필터링

````python
datas = np.array([-10, -2, 80, 999, 30, 120])
datas
````

````python
datas[(datas >= 0) & (datas <= 100)]
````
